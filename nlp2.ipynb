Information extraction using TF-IDF

```python
import nltk
import math
from collections import Counter

# Download NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# ---------------------------------------------------
# DOCUMENT COLLECTION
# ---------------------------------------------------

documents = [
    "Artificial Intelligence is transforming healthcare.",
    "Machine Learning is used for disease prediction.",
    "Artificial Intelligence and Machine Learning are important technologies.",
    "Healthcare organizations use Artificial Intelligence for medical diagnosis."
]

# ---------------------------------------------------
# NLTK PREPROCESSING
# ---------------------------------------------------

stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()

processed_documents = []

for document in documents:

    # Convert to lowercase
    document = document.lower()

    # Tokenization
    tokens = word_tokenize(document)

    # Remove stop words and non-alphabetic words
    tokens = [
        word for word in tokens
        if word.isalpha() and word not in stop_words
    ]

    # Stemming
    tokens = [stemmer.stem(word) for word in tokens]

    processed_documents.append(tokens)

print("PROCESSED DOCUMENTS")
print("=" * 50)

for i, doc in enumerate(processed_documents):
    print("Document", i + 1, ":", doc)


# ---------------------------------------------------
# CREATE VOCABULARY
# ---------------------------------------------------

vocabulary = set()

for document in processed_documents:
    vocabulary.update(document)

vocabulary = sorted(vocabulary)

print("\nVOCABULARY")
print("=" * 50)
print(vocabulary)


# ---------------------------------------------------
# TF CALCULATION
# ---------------------------------------------------

def calculate_tf(document):
    word_count = Counter(document)
    total_words = len(document)

    tf = {}

    for word in word_count:
        tf[word] = word_count[word] / total_words

    return tf


# ---------------------------------------------------
# IDF CALCULATION
# ---------------------------------------------------

def calculate_idf(word, documents):

    number_of_documents = len(documents)

    document_frequency = sum(
        1 for document in documents
        if word in document
    )

    return math.log(
        number_of_documents / document_frequency
    )


# ---------------------------------------------------
# TF-IDF CALCULATION
# ---------------------------------------------------

print("\nTF-IDF VALUES")
print("=" * 50)

for i, document in enumerate(processed_documents):

    tf = calculate_tf(document)

    tfidf = {}

    for word in document:

        idf = calculate_idf(word, processed_documents)

        tfidf[word] = tf[word] * idf

    print("\nDocument", i + 1)

    # Sort words according to TF-IDF value
    sorted_words = sorted(
        tfidf.items(),
        key=lambda x: x[1],
        reverse=True
    )

    for word, score in sorted_words:
        print(f"{word:15} : {score:.4f}")


# ---------------------------------------------------
# EXTRACT IMPORTANT KEYWORDS
# ---------------------------------------------------

print("\nIMPORTANT KEYWORDS")
print("=" * 50)

for i, document in enumerate(processed_documents):

    tf = calculate_tf(document)

    tfidf = {}

    for word in document:
        idf = calculate_idf(word, processed_documents)
        tfidf[word] = tf[word] * idf

    # Select top 3 keywords
    top_keywords = sorted(
        tfidf.items(),
        key=lambda x: x[1],
        reverse=True
    )[:3]

    print(
        f"Document {i + 1}:",
        [word for word, score in top_keywords]
    )
```

```python
import nltk

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# --------------------------------------------------
# STEP 1: DOCUMENT COLLECTION
# --------------------------------------------------

documents = [
    "Artificial Intelligence is used in healthcare.",
    "Machine Learning helps computers learn from data.",
    "Natural Language Processing helps computers understand human language.",
    "Cybersecurity protects computer networks from attacks.",
    "Deep Learning is used for image recognition and speech processing."
]


# --------------------------------------------------
# STEP 2: NLTK PREPROCESSING
# --------------------------------------------------

stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()


def preprocess(text):

    # Convert to lowercase
    text = text.lower()

    # Tokenization
    tokens = word_tokenize(text)

    # Remove stop words and punctuation
    tokens = [
        word for word in tokens
        if word.isalpha() and word not in stop_words
    ]

    # Stemming
    tokens = [stemmer.stem(word) for word in tokens]

    # Convert tokens back into a sentence
    return " ".join(tokens)


# Preprocess all documents
processed_documents = [
    preprocess(document)
    for document in documents
]


# --------------------------------------------------
# STEP 3: GET QUERY FROM USER
# --------------------------------------------------

query = input("Enter your search query: ")

processed_query = preprocess(query)


# --------------------------------------------------
# STEP 4: TF-IDF REPRESENTATION
# --------------------------------------------------

vectorizer = TfidfVectorizer()

# Combine documents and query
all_text = processed_documents + [processed_query]

tfidf_matrix = vectorizer.fit_transform(all_text)


# --------------------------------------------------
# STEP 5: CALCULATE COSINE SIMILARITY
# --------------------------------------------------

query_vector = tfidf_matrix[-1]

document_vectors = tfidf_matrix[:-1]

similarity_scores = cosine_similarity(
    query_vector,
    document_vectors
)[0]


# --------------------------------------------------
# STEP 6: RANK DOCUMENTS
# --------------------------------------------------

ranked_documents = similarity_scores.argsort()[::-1]


# --------------------------------------------------
# STEP 7: DISPLAY SEARCH RESULTS
# --------------------------------------------------

print("\nINFORMATION RETRIEVAL RESULTS")
print("=" * 60)

for rank, index in enumerate(ranked_documents, start=1):

    print("\nRank:", rank)

    print(
        "Similarity Score:",
        round(similarity_scores[index], 4)
    )

    print(
        "Document:",
        documents[index]
    )
```

Relation Extraction

```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk

# ------------------------------------------------
# INPUT TEXT
# ------------------------------------------------

text = """
Google developed Gemini in California.
Microsoft created Copilot in the United States.
Elon Musk founded SpaceX in California.
"""

# ------------------------------------------------
# PROCESS EACH SENTENCE
# ------------------------------------------------

sentences = nltk.sent_tokenize(text)

print("RELATION EXTRACTION USING NLTK")
print("=" * 60)

for sentence in sentences:

    print("\nSentence:", sentence)

    # Tokenization
    tokens = word_tokenize(sentence)

    # POS tagging
    pos_tags = pos_tag(tokens)

    # Named Entity Recognition
    tree = ne_chunk(pos_tags)

    # ------------------------------------------------
    # Extract named entities
    # ------------------------------------------------

    entities = []

    for subtree in tree:

        if hasattr(subtree, 'label'):

            entity_name = " ".join(
                word for word, tag in subtree
            )

            entity_type = subtree.label()

            entities.append(
                (entity_name, entity_type)
            )

    print("Entities:", entities)

    # ------------------------------------------------
    # Simple relation extraction
    # ------------------------------------------------

    relation_words = [
        "developed",
        "created",
        "founded",
        "works",
        "owns",
        "acquired",
        "invented",
        "launched"
    ]

    found_relation = None

    for word, tag in pos_tags:

        if word.lower() in relation_words:

            found_relation = word.lower()
            break

    # ------------------------------------------------
    # Display relation
    # ------------------------------------------------

    if found_relation and len(entities) >= 2:

        subject = entities[0][0]
        object_entity = entities[1][0]

        print("\nExtracted Relation:")
        print(subject, "→", found_relation, "→", object_entity)

    else:

        print("\nNo relation found.")
```
