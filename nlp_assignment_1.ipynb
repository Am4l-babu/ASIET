```python
import nltk
import string
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

```python
text = """
My name is Amal Babu, and I am from Chalakudy, Kerala. I completed my Diploma in Mechanical Engineering from Thiagarajar Polytechnic College, and I am currently pursuing a B.Tech in Computer Science and Engineering with a specialization in Artificial Intelligence at Adi Shankara Institute of Engineering and Technology. I come from a supportive family that has always encouraged me to learn new things and explore technology. Over the years, I have worked on several projects related to embedded systems, IoT, Edge AI, and digital twin technology, which have helped me improve my practical and problem-solving skills. I enjoy participating in hackathons and technical events because they give me opportunities to learn from others and apply my knowledge to real-world challenges. In my free time, I like building electronic projects, designing 3D-printed models, experimenting with new technologies, and reading about artificial intelligence and robotics. My strengths include quick learning, creativity, teamwork, leadership, and analytical thinking. One area I am working to improve is time management because I often spend extra time refining my projects to achieve the best possible result. My goal is to become an engineer who develops intelligent embedded systems that make a positive impact on society.
"""

print(text)
```

```python
sentences = sent_tokenize(text)

print("Sentence Tokenization\n")

for sentence in sentences:
    print(sentence)
```

```python
words = word_tokenize(text)

print(words)
```

```python
lowercase = [word.lower() for word in words]

print(lowercase)
```

```python
no_punctuation = [word for word in lowercase if word not in string.punctuation]

print(no_punctuation)
```

```python
stop_words = set(stopwords.words("english"))

filtered_words = [word for word in no_punctuation if word not in stop_words]

print(filtered_words)
```

```python
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print(stemmed_words)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print(lemmatized_words)
```

```python
clean_text = " ".join(lemmatized_words)

print(clean_text)
```
